```
import torch
import numpy as np
```

```
t1 = torch.tensor([1, 2, 3])
t2 = torch.tensor([2, 3, 4])
# 运算符重载，非原地
print(t1 + t2)
```

```
# 带下划线，原地修改
t1.add_(t2)
print(t1)
```

```
# 其他运算
print(t1 ** 2)
print(t1.sqrt())
print(t1.logit(2))
```

```
# 矩阵乘法
t2 = torch.rand(3, 4)
t3 = torch.rand(4, 2)
print(t2 @ t3)
```

```
# 哈达玛积（逐元素相乘）
t2 = torch.rand(3, 4)
t3 = torch.rand(3, 4)
print(t2, t3, t2 * t3)
```
