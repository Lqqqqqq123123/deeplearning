```python
import numpy as np
# 用列表去创建
arr = np.array([1, 2, 3, 4, 5])
print(arr)
```

```python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])
print(arr.shape)
```

```python
# 内置函数创建
arr = np.zeros((2, 3))
print(arr)
arr = np.ones((2, 3), dtype=np.int32)
print(arr)
arr = np.full((2, 3), 666)
print(arr)
arr = np.diag([1, 2, 3, 4])
print(arr)
```

```python
# 序列创建
arr = np.arange(0, 10, 2)
print(arr)
arr = np.arange(0, 10)
print(arr)
arr = np.linspace(0, 10, 5)
print(arr)
arr = np.random.rand(2, 3)
print(arr)
arr = np.random.randn(2, 3)
print(arr)
```
